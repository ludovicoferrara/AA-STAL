# Person–Object Association

This stage reads the existing object-tracking JSON files and produces frame-level and window-level person–object associations using bounding-box overlap, proximity, and temporal consistency. It does not rerun object detection or SAM2.

```python
from google.colab import drive
drive.mount('/content/drive')

%cd /content/drive/MyDrive/AA-STAL/data_pipeline
```

```python
import os
from person_object_association import (
    AssociationConfig,
    process_directory,
    render_directory,
)

PATH_DATA_ROOT = '/content/drive/MyDrive/AA-STAL/data_pipeline/DATA_ROOT'
PATH_TRACKING = os.path.join(PATH_DATA_ROOT, 'video_general_obj_det_finished')
PATH_ASSOCIATIONS = os.path.join(PATH_DATA_ROOT, 'person_object_associations')

CONFIG = AssociationConfig(
    person_bbox_expansion=0.10,
    overlap_weight=0.75,
    proximity_weight=0.25,
    max_distance_ratio=0.25,
    minimum_frame_score=0.30,
    window_size=60,
    stride=30,
    minimum_support_ratio=0.50,
    minimum_consecutive_frames=5,
)

processed, skipped = process_directory(PATH_TRACKING, PATH_ASSOCIATIONS, CONFIG)
print(f'Association completed: {processed} videos processed, {skipped} skipped.')
```

## Visual validation

Generate diagnostic copies of sampled frames. People are shown in red, temporally accepted objects in green, and person–object links in yellow. Set `INCLUDE_FRAME_ONLY_ASSOCIATIONS=True` to also inspect rejected frame-level candidates in orange.

```python
PATH_FRAMES = os.path.join(PATH_DATA_ROOT, 'Videos_crop_decode')
PATH_VISUALIZATIONS = os.path.join(
    PATH_DATA_ROOT, 'person_object_association_visualizations'
)
VISUALIZATION_FRAME_STEP = 30
INCLUDE_FRAME_ONLY_ASSOCIATIONS = False

rendered, skipped = render_directory(
    tracking_directory=PATH_TRACKING,
    association_directory=PATH_ASSOCIATIONS,
    frames_root=PATH_FRAMES,
    output_root=PATH_VISUALIZATIONS,
    frame_step=VISUALIZATION_FRAME_STEP,
    include_frame_only_associations=INCLUDE_FRAME_ONLY_ASSOCIATIONS,
)
print(f'Visualization completed: {rendered} frames rendered, {skipped} videos skipped.')
```
